```python
from src.config import get_con
con = get_con(read_only=True)
```

```python
con.execute("""
    SELECT get_repo_origin(repo_url) as origin, columns.column_base_type, COUNT(*) as cnt
    FROM repos
    JOIN queries ON queries.repo_id = repos.id
    JOIN column_usages_unnested as cu ON cu.query_id = queries.id
    JOIN columns ON columns.id = cu.column_id
    WHERE usage_type = 'JOIN_KEY'
    GROUP BY ALL
    ORDER BY origin, cnt DESC
""").fetchdf()
```
